## 2. Training and Evaluation

We train a YOLOv8 Nano model using pretrained weights.

```python
from ultralytics import YOLO
import os

DATASET_PATH = os.path.abspath("YOLO_COCO")
YAML_PATH = os.path.join(DATASET_PATH, "data.yaml")

# Load pretrained YOLOv8 model
model = YOLO("yolov8n.pt")

# Train model
results = model.train(
    data=YAML_PATH,
    epochs=100,
    imgsz=640,
    batch=16,
    workers=8,
    cache='disk',
    optimizer='Adam',
    project="yolo_results",
    name="sign_type_run"
)
```

```python
from ultralytics import YOLO
import os

# 1. Setup paths
DATASET_PATH = os.path.abspath("YOLO_COCO")
YAML_PATH = os.path.join(DATASET_PATH, "data.yaml")

# 2. Load your best trained model weights
model_path = "yolo_results/sign_type_run/weights/best.pt"
model = YOLO(model_path)

# 3. Run validation on the 'val' split
metrics = model.val(
    data=YAML_PATH,
    split='val',
    imgsz=640,
    batch=16,
    device='cpu',
)

# 4. Extract and print the key quantitative scores for your report
print(f"Mean Average Precision (mAP) @ 50: {metrics.box.map50:.4f}")
print(f"Mean Average Precision (mAP) @ 50-95: {metrics.box.map:.4f}")
print(f"Fitness Score (Combined): {metrics.fitness:.4f}")
```

## 4. YOLOv8 Detection Visualisation

```python
import cv2
import os
import matplotlib.pyplot as plt
from ultralytics import YOLO
from pathlib import Path

# 1. Setup Paths
DATASET_PATH = os.path.abspath("YOLO_COCO")
TEST_IMAGES_DIR = Path(DATASET_PATH) / "images" / "test"
OUTPUT_VIS_DIR = Path("visualizations/yolov8_results")
OUTPUT_VIS_DIR.mkdir(parents=True, exist_ok=True)

# 2. Load trained model
model = YOLO("yolo_results/sign_type_run/weights/best.pt")

# 3. Process Test Images
print(f"Running visualization on images in: {TEST_IMAGES_DIR}")
test_images = list(TEST_IMAGES_DIR.glob("*.jpg"))

for image_path in test_images[:10]:
    # Run inference
    results = model(image_path, conf=0.25)[0]
    
    # Load image for plotting
    img = cv2.imread(str(image_path))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    plt.figure(figsize=(12, 8))
    plt.imshow(img)
    plt.title(f"Detections for {image_path.name}")
    plt.axis("off")
    
    # Draw boxes manually for higher quality control in report
    for box in results.boxes:
        # Get coordinates, class, and confidence
        x1, y1, x2, y2 = box.xyxy[0].cpu().numpy()
        cls_id = int(box.cls[0])
        conf = box.conf[0].item()
        label = model.names[cls_id]
        
        # Add Rectangle
        plt.gca().add_patch(
            plt.Rectangle((x1, y1), x2 - x1, y2 - y1, 
                         fill=False, edgecolor='lime', linewidth=3)
        )
        
        # Add Label text
        plt.text(x1, y1 - 10, f"{label} {conf:.2f}", 
                 color='white', fontsize=12, fontweight='bold',
                 bbox=dict(facecolor='lime', alpha=0.8, edgecolor='none'))

    save_path = OUTPUT_VIS_DIR / f"val_result_{image_path.name}"
    plt.savefig(save_path, bbox_inches='tight', pad_inches=0.1)
    plt.close()

print(f"Done! {len(test_images[:10])} images saved to {OUTPUT_VIS_DIR}")
```
